# 03 - YOLO Training
Train the object detection model on the prepared dataset using Ultralytics YOLOv8.

```python
!pip install ultralytics
```

```python
from ultralytics import YOLO
import os
```

## Load Model

```python
model = YOLO("yolov8n.pt")  # Use nano model for faster training
```

## Train

```python
results = model.train(
    data="../config/dataset.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    project="../outputs/training",
    name="smart_safety_detector"
)
```

## Save best weights

```python
import shutil
best_weight = os.path.join("../outputs/training", "smart_safety_detector", "weights", "best.pt")
if os.path.exists(best_weight):
    shutil.copy(best_weight, "../models/best.pt")
    print("Copied best.pt to models/")
```
