Explore the workplace safety dataset structure, distributions, and visualize sample images and annotations.

```python
import os
import glob
import cv2
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import yaml
```

## Configuration & Validation

```python
with open("../config/dataset.yaml", "r") as f:
    config = yaml.safe_load(f)

print("Dataset configuration loaded:")
print(config)

DATA_DIR = config["path"]
```

## Image and Label Counts

```python
train_images = glob.glob(os.path.join(DATA_DIR, config["train"], "*.jpg"))
val_images = glob.glob(os.path.join(DATA_DIR, config["val"], "*.jpg"))
print(f"Total train images: {len(train_images)}")
print(f"Total validation images: {len(val_images)}")
```
