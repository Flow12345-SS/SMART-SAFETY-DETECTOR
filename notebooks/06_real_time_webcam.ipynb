# 06 - Real-time Webcam Inference
Use cv2.VideoCapture to run real-time inference using the trained model.

```python
import cv2
from ultralytics import YOLO
```

```python
model = YOLO("../models/best.pt")
cap = cv2.VideoCapture(0)
```

```python
while cap.isOpened():
    success, frame = cap.read()
    if not success:
        break
    results = model(frame, conf=0.5)
    annotated_frame = results[0].plot()
    cv2.imshow("Smart Safety Detector", annotated_frame)
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break
cap.release()
cv2.destroyAllWindows()
```
