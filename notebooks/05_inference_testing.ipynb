# 05 - Inference Testing
Test the trained YOLO model on static images and a video file.

```python
import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO
```

```python
model = YOLO("../models/best.pt")
```

## Predict on Image

```python
results = model.predict(source="../data/images/test/sample.jpg", conf=0.4)
res_plotted = results[0].plot()
plt.imshow(cv2.cvtColor(res_plotted, cv2.COLOR_BGR2RGB))
plt.show()
```
